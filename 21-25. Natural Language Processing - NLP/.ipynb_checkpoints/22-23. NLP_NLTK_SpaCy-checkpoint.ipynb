```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')

# Downloads the full popular package collection
nltk.download('popular')
```

```python
text = "Natural Language Processing is interesting."

tokens = nltk.word_tokenize(text)

print(tokens)
```

```python
# spaCy - Particularly used for production-oriented NLP pipelines.
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")
doc = nlp("Apple is opening a new office in London.")

for token in doc:
    print(token.text, token.pos_)   # Tokens = Words, POS = Part of Speech (Grammar)
```

```python
# Getting POS tags
```

```python
import nltk

# Download tokenizers and the POS tagger models
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')


text = "Ali is running in Lahore."

tokens = nltk.word_tokenize(text)
tags = nltk.pos_tag(tokens)

print(tags)
```

```python
# Python with SpaCy - Output: Dependency Parse Tree
```

```python
import spacy

# 1. Load the English spaCy language model
nlp = spacy.load("en_core_web_sm")

# 2. Define your input text in a separate variable
text = "Ali bought a new car."

# 3. Process the text through the spaCy pipeline
doc = nlp(text)

# 4. Print token text, dependency label name, and head token
for token in doc:
    print(token.text, token.dep_, token.head.text)

# Here:
# nsubj = nominal subject
# dobj = direct object
# det = determiner
# amod = adjectival modifier
# ROOT = main/root word
```
