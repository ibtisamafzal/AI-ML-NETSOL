```python
# Python with SpaCy - Output: Dependency Parse Tree
```

```python
import spacy

# 1. Load the English spaCy language model
nlp = spacy.load("en_core_web_sm")

# 2. Define your input text in a separate variable
text = "Ali bought a new car."

# 3. Process the text through the spaCy pipeline
doc = nlp(text)

# 4. Print token text, dependency label name, and head token
for token in doc:
    print(token.text, token.dep_, token.head.text)

# Here:
# nsubj = nominal subject
# dobj = direct object
# det = determiner
# amod = adjectival modifier
# ROOT = main/root word
```

```python
# Stemming - Finds the base word from similar words
```

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()

words = [
    "play",
    "playing",
    "played",
    "plays"
]

for word in words:
    print(word, "->", stemmer.stem(word))
```

```python
# Another Example - Stemming don't always give you correct output
```

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()

words = [
    "studies",
    "studying",
    "study"
]

for word in words:
    print(word, "->", stemmer.stem(word))
```

```python
# Lemmatization
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = ["cars","dogs","running"]

for word in words:
    print(word, "->", lemmatizer.lemmatize(word))
```

```python
# NER - Named Entity Recognition
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

text = "Ali works at Google in Lahore."

doc = nlp(text)

for entity in doc.ents:
    print(entity.text, entity.label_)
```

```python
# WordNet
# WordNet is a lexical database that groups words into sets of synonyms called synsets, each representing a distinct concept.
```

```python
from nltk.corpus import wordnet
synsets = wordnet.synsets("car")

for synset in synsets:
    print(synset)

# Print definitions for each synset
for syn in wordnet.synsets("car"):
    print(syn.name())
    print(syn.definition())
    print()
```

```python
# Regular Expressions
```

```python
import re

# For Digits
text = "I have 25 apples and 10 oranges."
numbers = re.findall(r"\d+", text)
print(numbers)

# For Email:
text1 = "Contact me at: user@example.com"
emails = re.findall(r'[\w\.-]+@[\w\.-]+\.\w+', text1)
print(emails)

# For URL:
text2 = "Verify https://example.com"
urls = re.findall(r'https?://\S+', text2)
print(urls)
```

```python
# Bag of Words (BoW)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
documents = [
    "I love Python",
    "I love machine learning",
    "Python is powerful"
]

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(documents)
print(vectorizer.get_feature_names_out())
print(X.toarray())
```

```python
# Feature Extraction
```

```python
from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "machine learning is powerful",
    "machine learning is useful"
]

vectorizer = CountVectorizer(
    ngram_range = (1,2)
)

X = vectorizer.fit_transform(documents)
print(vectorizer.get_feature_names_out())
```

```python
# TF-IDF
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

documents = [
    "python is great",
    "python is powerful",
    "Machine learning is powerful"
]

vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(documents)
print(vectorizer.get_feature_names_out())
print(X.toarray())
```

```python
# Cosine Similarity
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

documents = [
    "I love machine learning",
    "Machine learning is amazing",
    "I like cooking food"
]

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(documents)

similarity = cosine_similarity(X)
print(similarity)
```
