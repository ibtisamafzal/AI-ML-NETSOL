```python
# Age Check If-Else Nested Statement
```

```python
age = 18
has_license = True
```

```python
if age >=18:
    if has_license:
        print("You are an adult and can drive")
else:
    print("Minor, you need a license")
```

```python
# OR it can be done by this statement
```

```python
if age >= 18 and has_license:
    print("Adult and can drive")
```

```python
# Even-Odd If-Else Statement
```

```python
number = 2
```

```python
if number % 2 == 0:
    print("Even")
else:
    print("Odd")
```

```python
# Elif-Statement
```

```python
marks = 80
```

```python
if marks >=90:
    Grade = "A"
elif marks >=70:
    Grade = "B"
elif marks >=50:
    Grade = "B"
elif marks >=30:
    Grade = "B"
else:
    Grade = "F"

print(Grade)
```
