```python
import spacy
```

```python
# Syntax - Sentence Structure
```

```python
nlp = spacy.load("en_core_web_sm")
doc = nlp("The cat chased the mouse.")

for token in doc:
    print(token.text, token.pos_, token.dep_, token.head.text)
```
