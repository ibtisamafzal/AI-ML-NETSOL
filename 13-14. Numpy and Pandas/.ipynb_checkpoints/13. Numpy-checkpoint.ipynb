```python
# Numpy Array
import numpy as np
numbers = np.array([10,20,30,40])
print(numbers)
```

```python
# NumPy Array vs Python List (Same syntax but different behavior)
```

```python
# NumPy Array: Element-Wise Vectorized Operation
# In NumPy, numbers * 2 performs vectorized arithmetic (broadcasting).
# It applies scalar multiplication to every individual element inside the array.
```

```python
numbers = np.array([10,20,30,40])
print(numbers * 2)
```

```python
# Python Standard List: Sequence Replication
# In standard Python, lists are generic containers, not mathematical vectors.
# The * operator on a sequence defines sequence repetition.
# It duplicates and concatenates the elements rather than modifying the values within.
```

```python
numbers = [10,20,30,40]
print (numbers * 2) 
```

```python
# 2-D Array (similar to a table or matrix)
```

```python
arr = np.array([
    [10,20,30],
    [40,50,60]
])

print(arr)
```

```python
# Creating Arrays using Built-in Methods
```

```python
# Creates an array with zeros
arr = np.zeros(5)
print(arr)


# 2D Array
arr = np.zeros((2,3))
print(arr)


# Creates an array containing ones
arr = np.ones(5)
print(arr)


# Create an array filled with a specific value
arr = np.full(5,7)
print(arr)


# Range
arr = np.arange(1,10)
print(arr)


# Create evenly spaced numbers between two values (useful in plotting and calculations)
arr = np.linspace(0,10,5)
print(arr)
```

```python
# Creating Arrays using Random Numbers
```

```python
# Random Array
arr = np.random.rand(5)  # -> Numpy = Package, random = library, rand = function
print(arr)


# Random Integers
arr = np.random.randint(1,100,5)
print(arr)


# Random 2D Array
arr = np.random.randint(1,10, size=(3,4))
print(arr)
```

```python
# NumPy Aggregation Methods
```

```python
arr = np.array([10,50,20,80,30])
print(arr.max())
print(arr.min())

print(arr.argmax())  # returns index of the largest value
```
