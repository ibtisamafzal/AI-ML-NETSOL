```python
# Multiple Conditions
```

```python
data = {
    "Name": ["Ali", "Ahmed", "Sara"],
    "Age": [25, 30, 28],
    "Salary": [50000, 70000, 60000]
}
```

```python
# 1. Convert the dictionary into a DataFrame
df = pd.DataFrame(data)

# 2. Age > 25 AND salary > 60000

result_and = df[(df["Age"] > 25) & (df["Salary"] > 60000)]
print(result_and)
```
