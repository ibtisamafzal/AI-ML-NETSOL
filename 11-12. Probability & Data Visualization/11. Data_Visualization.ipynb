```python
# Correlation Coefficient
```

```python
import pandas as pd

df = pd.DataFrame({
    "hours":[1,2,3,4,5],
    "score":[45,52,61,70,78]
})

print(df["hours"].corr(df["score"]))
```

```python
# Univariate Plot - Histogram/ Boxplot/ BarChart
```

```python
import matplotlib.pyplot as plt

ages = [18,19,20,21,22,400,401,402,403, 2000,2001,2002,2003,2004]

# Histogram (Skewness, Possible Outliers)
plt.hist(ages)
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()

# Boxplot (Useful for Median, Quartiles, Spread, Outliers)
plt.boxplot(ages)
plt.show()

# Bar Chart
cities = ["Lahore", "Karachi", "Islamabad"]
customers = [500, 700, 300]

plt.bar(cities, customers)
plt.show()
```

```python
# Bivariate Plot - Scatter Plot
```

```python
import matplotlib.pyplot as plt

hours = [1,2,3,4,5]
scores = [45,52,61,70,78]

# Scatter PLot
plt.scatter(hours, scores)
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.show()
```

```python
# Multivariate Plot - Pair Plot/ 3D Scatter Plot
```

```python
import seaborn as sns

# Pair Plot
sns.pairplot(df)
plt.show()

x = [3119,956,602,1673,1504,686,1593]
y = [1,5,4,2,5,2,5]
z = [1,2,3,4,5,6,7]

# 3D Scatter Plot
fig = plt.figure()
ax = fig.add_subplot(projection="3d")

ax.scatter(x,y,z)
ax.set_xlabel("Area")
ax.set_ylabel("Bedrooms")
ax.set_zlabel("Price")

plt.show()
```
