```python
# Correlation
```

```python
import pandas as pd

df = pd.DataFrame({
    "hours":[1,2,3,4,5],
    "score":[45,52,61,70,78]
})

correlation = df["hours"].corr(df["score"])

print(correlation)
```

```python
# Conditional Probability
```

```python
import pandas as pd

df = pd.DataFrame({
    "gender":["M","M","M","F","F"],
    "purchased": [1,0,1,1,0]
})

male_customers = df[df["gender"] == "M"]

probability = male_customers["purchased"].mean()
print(probability)
```
