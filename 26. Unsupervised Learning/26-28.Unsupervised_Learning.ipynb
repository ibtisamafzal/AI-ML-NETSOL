```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler

# Generate Unlabeled Dataset
X,_ = make_blobs(
    n_samples=300,
    centers=3,
    cluster_std=1.2,
    random_state=42
)

# Scale Features
X = StandardScaler().fit_transform(X)

# Train the model
model = KMeans(
    n_clusters=3,
    init="k-means++",
    n_init=10,
    random_state=42
)

labels = model.fit_predict(X)
print("Cluster Labels:", labels[:10])
print("WCSS / Inertia:", model.inertia_)

plt.scatter(X[:, 0], X[:, 1], c=labels)
plt.scatter(
    model.cluster_centers_[:,0],
    model.cluster_centers_[:,1],
    marker="X",
    s=200
)

print(X[0][0])
print(labels)


plt.title("K-Means Clustering")
plt.show()
```

```python
# Hierarchical Clustering
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler

# Generate Unlabeled Dataset
X,_ = make_blobs(
    n_samples=200,
    centers=3,
    random_state=42
)

# Scale Features
X = StandardScaler().fit_transform(X)

model = AgglomerativeClustering(
    n_clusters=3,
    linkage="ward"
)

labels = model.fit_predict(X)

plt.scatter(X[:, 0], X[:, 1], c=labels)
plt.title("Hierarchical Clustering")
plt.show()
```

```python
# DBSCAN Clustering (Density-Based Spatial Clustering) - Works with Noise 
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons
from sklearn.preprocessing import StandardScaler

# Generate Unlabeled Dataset
X,_ = make_moons(
    n_samples=300,
    noise=0.06,
    random_state=42
)

# Scale Features
X = StandardScaler().fit_transform(X)

model = DBSCAN(
    eps=0.25,
    min_samples=5
)

labels = model.fit_predict(X)

print("Labels:", set(labels))
print("Noise points:", sum(labels==-1))
plt.scatter(X[:, 0], X[:, 1], c=labels)
plt.title("DBSCAN Clustering")
plt.show()
```

```python
# Gaussian Mixture Model (GMM)
```

```python
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler

# Generate Unlabeled Dataset
X,_ = make_blobs(
    n_samples=300,
    centers=3,
    cluster_std=1.5,
    random_state=42
)

# Scale Features
X = StandardScaler().fit_transform(X)

model = GaussianMixture(
    n_components=3,
    covariance_type="full",
    random_state=42
)

model.fit(X)

labels = model.predict(X)
probabilities = model.predict_proba(X)
print("Negative log-likelihood:", -model.score(X))
print("First five probability assignments:")
print(probabilities[:5])

plt.scatter(X[:, 0], X[:, 1], c=labels)
plt.title("Gaussian Mixture Model (GMM)")
plt.show()
```

```python
# Principal Component Analysis (PCA)
```

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Ignore the target labels
X = load_iris().data


# Scale Features
X_scaled = StandardScaler().fit_transform(X)

pca = PCA(
    n_components=2)

X_reduced = pca.fit_transform(X_scaled)

print("Original shape:", X.shape)
print("Reduced shape:", X_reduced.shape)
print("Variance explained:", pca.explained_variance_ratio_)
print("Total variance retained:", pca.explained_variance_ratio_.sum())


plt.scatter(X_reduced[:,0], X_reduced[:, 1])
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA: Two-Dimensional Representation")
plt.show()
```

```python
# Autoencoders
```

```python
import torch
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from torch import nn
from torch.utils.data import DataLoader, TensorDataset

# Prepare unlabeled input data
X = load_iris().data
X = StandardScaler().fit_transform(X)

X_tensor = torch.tensor(X, dtype=torch.float32)
loader = DataLoader(
    TensorDataset(X_tensor), batch_size=16, shuffle=True  # Fixed typo: DataLaoder -> DataLoader
)


class Autoencoder(nn.Module):

    def __init__(self):  # Fixed typo: space added in `def __init__`
        super().__init__()  # Fixed indentation

        self.encoder = nn.Sequential(
            nn.Linear(4, 3),
            nn.ReLU(),  # Fixed typo: nn.ReLu -> nn.ReLU
            nn.Linear(3, 2),
        )

        self.decoder = nn.Sequential(
            nn.Linear(2, 3),
            nn.ReLU(),  # Fixed typo: nn.ReLu -> nn.ReLU
            nn.Linear(3, 4),
        )

    def forward(self, x):
        latent = self.encoder(x)
        reconstructed = self.decoder(latent)
        return reconstructed


model = Autoencoder()
loss_fn = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

for epoch in range(100):
    total_loss = 0

    for (batch,) in loader:
        optimizer.zero_grad()
        reconstructed = model(batch)
        loss = loss_fn(reconstructed, batch)
        loss.backward()
        optimizer.step()
        total_loss += loss.item() * len(batch)

    if (epoch + 1) % 20 == 0:
        print(f"Epoch {epoch + 1}, Loss: {total_loss / len(X_tensor):.4f}")

with torch.no_grad():
    latent_features = model.encoder(X_tensor)
    reconstructed = model(X_tensor)

    print("Latent shape:", latent_features.shape)
    print(
        "Final reconstruction MSE:", loss_fn(reconstructed, X_tensor).item()
    )
```

```python
# Association Rule Learning
```

```python
import pandas as pd
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules

transactions = [
    ["Bread","Butter","Milk"],
    ["Bread","Butter"],
    ["Bread","Milk"],
    ["Butter","Milk"],
    ["Bread","Butter","Milk"],
    ["Bread","Butter"]
]

# Convert transactions to one-hot encoded data
items = sorted({
    item
    for transaction in transactions
    for item in transaction
})

df = pd.DataFrame([
    {
    item:item in transaction
    for item in items
    }
    for transaction in transactions
])

# Find Frequent itemsets
frequent_itemsets = apriori(
    df,
    min_support=0.3,
    use_colnames=True
)

print(frequent_itemsets)



```
