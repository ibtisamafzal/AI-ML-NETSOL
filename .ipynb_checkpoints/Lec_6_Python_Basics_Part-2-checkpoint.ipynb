```python
# LAMBDA Function
```

```python
students = [("Ali", 80), ("Ibtisam",95),("Ahmad",70)]

students.sort(key=lambda student: student[1])

print(students)
```

```python
# MAP - Applies a function to every element of an iterable and returns an iterator.
```

```python
numbers = [1,2,3,4]

result = map(lambda x:x**2, numbers)

# Return Object of Map
print(result)

# Type-cast to get output in your desired structure
print(list(result))
```

```python
# MAP for Built-in Square Function
```

```python
numbers = [1, 2, 3, 4, 5, 6]

def is_even(x):
    return x % 2 == 0

evens = list(filter(is_even, numbers))
print(evens)
```

```python
numbers = [1, 2, 3, 4, 5, 6]

result = filter(lambda x:x%2 ==0,  numbers)

print(list(result))
```

```python
# Inner/ Nested Functions
```

```python
def outer():
    
    def inner():
        print("Hello from inner")
    inner()
    
    print("Hello from Outer")
outer()    
```

```python
# Closure Functions
```

```python
def multiplier(factor):
    
    def multiply(number):
        return number * factor
    
    return multiply
    
double = multiplier(2)
triple = multiplier(3)

print(double(10))
print(triple(20))
```
