```python
# Abstraction
```

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):
        print("Bark")

dog = Dog()
dog.sound()

# animal = Animal()
```

```python
# Abstract Class - Serve as a base/ template for subclasses.
```

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius ** 2
        

circle = Circle(5)
print(circle.area())
```

```python
# Inheritance
```

```python
class Animal(ABC):      # Parent Class
    def eat(self):
        print("Eating")

class Dog(Animal):      # Child Class
    def bark(self):
        print("Barking")

dog = Dog()
dog.eat()
dog.bark()
```

```python
# Method Overloading
```

```python
class Calculator:
    def add(self, a, b, c=0):    # Default argument is required, otherwise gives error
        return a + b + c

calculator = Calculator()
print(calculator.add(1, 2))     # 3
print(calculator.add(1, 2, 3))  # 6
```
