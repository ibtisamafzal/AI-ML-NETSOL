```python
# Class and Object
```

```python
class Car:
    def drive(self):     # Constructor
        print("Car is driving")
```

```python
car1 = Car()
car1.drive()
```

```python
class Student:
    def __init__(self, name, age):    # Constructor
        self.name = name
        self.age = age

std1 = Student("Ibtisam",20)
std2 = Student("Adnan",30)

print(std1.name)
print(std1.age)
print(std2.name)
print(std2.age)
```

```python
class Student:
    def __init__(self, name):     # Constructor
        self.name = name

    def introduce(self):
        print(f"My name is {self.name}")
        
std1 = Student("Ibtisam")
std2 = Student("Adnan")

std1.introduce()
std2.introduce()
```

```python
# Class Variable and Class Functions
```

```python
class Student:

    school = "ABC School"    # Class Variable
    
    def __init__(self, name):    # Constructor
        self.name = name          # Instance Variable

        
        if name == "Sufi":        # Class Function
            self.school = name

        
std1 = Student("Ibtisam")
std2 = Student("Adnan")
std3 = Student("Sufi")


print(std1.name)
print(std1.school)

print(std2.name)
print(std2.school)

print(std3.school)
```

```python
# Class Method - No need to create an object!
```

```python
class Student:

    school = "ABC School"    # Class Variable
    
    def __init__(self, name):  # Constructor
        self.name = name          # Instance Variable

        
    @classmethod
    def change_school(cls, new_school):
        cls.school = new_school


Student.change_school("XYZ School")
print(Student.school)
```

```python
# Static Methods
```

```python
class Math:

    @staticmethod
    def add(a,b):
        return a + b

# Static Method
print(Math.add(10,20))

# Object method
add1 = Math.add(20,40)
print(add1)
```

```python
# Encapsulation (Combining Data and the Methods)
```

```python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance


Ba1 = BankAccount(100)
print(Ba1.balance)
```

```python
# Name Mangling
```

```python
class Student:
    def __init__(self):
        self.__name = "Ali"

student = Student()
student._Student__name

student = Student()
print(student._Student__name)  # Accesses the name-mangled private attribute: Ali
```
