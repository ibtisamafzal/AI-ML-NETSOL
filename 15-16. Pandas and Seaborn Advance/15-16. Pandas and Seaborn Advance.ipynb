```python
# Data Plotting (Pandas and Seaborn)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.DataFrame({
    "month":["Jan","Feb","Mar","Apr","May"],
    "sales":[100,150,130,200,250],
    "profit":[20,35,25,50,70]
})
```

```python
# Area Plot

df.plot(
    x="month",
    y="sales",
    kind="area"
)

plt.show()
```

```python
# Bar Chart

df.plot(
    x="month",
    y="sales",
    kind="bar"
)
plt.show()
```

```python
# Horizontal Bar Chart

df.plot(
    x="month",
    y="sales",
    kind="barh"
)
plt.show()
```

```python
# Line Plot

df.plot(
    x="month",
    y="sales",
    kind="line",
    marker="*"
)
plt.show()
```

```python
# Histogram
df["sales"].plot(
    kind="hist",
    bins=5
)
plt.show()
```

```python
# Density Plot

df["sales"].plot(
    kind="density"
)
plt.show()
```

```python
# KDE Plot - Kernel Density Estimation
```

```python
# KDE Plot

df["sales"].plot(
    kind="kde"
)
plt.show()
```

```python
# Scatter Plot (Explore whether higher sales are associated with higher profits.

df.plot(
    x="sales",
    y="profit",
    kind="scatter"
)
plt.show()
```

```python
# Box Plot

df[["sales","profit"]].plot(kind="box")
plt.show()
```

```python
# Hexbin Plot

df = pd.DataFrame({
    "month":["Jan","Feb","Mar","Apr","May","June", "July"],
    "sales":[100,150,150, 130,200,250,250],
    "profit":[20,35,35,25,50,70,70]
})

df.plot(
    x="sales",
    y="profit",
    kind="hexbin",
    gridsize=10
)
plt.show()
```

```python
# Pie Chart

df.plot(
    x="sales",
    y="sales",
    kind="pie",
    labels=df["month"],
    legend=False
)
plt.show()
```

```python
# Seaborn - Built on top of matplotlib
```

```python
df = pd.DataFrame({
    "department":["CS","IT","CS","IT","CS","IT","CS","IT"],
    "gender":["M","F","M","F","F","M","M","F"],
    "marks":[85,90,78,95,88,70,92,82],
    "hours":[4,6,3,7,5,2,6,4],
    "age":[20,22,21,23,25,24,22,21]
})
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

sns.set_theme(style="whitegrid")
```

```python
sns.scatterplot(
    data=df,
    x="hours",
    y="marks"
)
plt.show()
```

```python
sns.displot(df["marks"])
plt.show()
```

```python
sns.jointplot(
    data=df,
    x="hours",
    y="marks",
    kind="scatter"
)
```

```python
# Scatter Plot
sns.jointplot(data=df, x="hours", y="marks", kind="scatter")

# Density Contours
sns.jointplot(data=df, x="hours", y="marks", kind="kde")

# Hexagonal Bins
sns.jointplot(data=df, x="hours", y="marks", kind="hex")
```

```python
# Pair Plot

sns.pairplot(
    df,
    vars=["marks","hours","age"],
    hue="department"
)
plt.show()
```

```python
# Rug Plot

sns.rugplot(
    data=df,
    x="marks"
)
plt.show()
```

```python
# Factor Plot (Not working) -> Cat Plot

sns.catplot(
    data=df,
    x="department",
    y="marks",
    kind="box"
)
plt.show()
```

```python
# Box Plot

sns.boxplot(
    data=df,
    x="department",
    y="marks"
)
plt.show()
```

```python
# Violin Plot

sns.violinplot(
    data=df,
    x="department",
    y="marks"
)
plt.show()
```

```python
sns.stripplot(
    data=df,
    x="department",
    y="marks",
    jitter=True
)
plt.show()
```

```python
# Swarm Plot

sns.swarmplot(
    data=df,
    x="department",
    y="marks"
)
plt.show()
```

```python
# Count Plot

sns.countplot(
    data=df,
    x="department"
)
plt.show()
```

```python
# Heatmap
```

```python
data = pd.DataFrame(
    [[10,20,30],
     [40,50,60],
     [70,80,90]
    ],

    index=["A","B","C"],
    columns=["X","Y","Z"]
)
```

```python
# Coolwarm Heatmap

sns.heatmap(
    data,
    annot=True,
    cmap="coolwarm"
)
plt.show()
```

```python
# Correlation Heatmap

correlation = df[["marks","hours","age"]].corr()
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)
plt.show()
```
