# Univariate Linear Regression with Gradient Descent

## Without Vectorization (plain Python loops)

```python
# Univariate Linear Regression with Gradient Descent - (Without Vectorization)
```

```python
# Training data: the true relationship is y = 2x, so we expect w ~ 2, b ~ 0
X = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]
```

```python
# Hyperparameters and initial parameters
w = 0.0               # weight (slope), starts at 0
b = 0.0               # bias (intercept), starts at 0
learning_rate = 0.01  # step size for each update
epochs = 100000        # number of full passes over the data

m = len(X)            # number of training examples
```

```python
for epoch in range(epochs):
    # Reset gradient accumulators at the start of every epoch
    dw = 0.0
    db = 0.0

    # Loop over every training example to accumulate gradients
    for i in range(m):
        y_pred = w * X[i] + b      # model prediction for example i
        
        error = y_pred - y[i]      # how far off the prediction is

        dw += error * X[i]         # partial derivative of loss w.r.t. w
        db += error                # partial derivative of loss w.r.t. b

    # Average the gradients over all examples
    dw = dw / m
    db = db / m

    # Gradient descent update: move opposite to the gradient
    w = w - learning_rate * dw
    b = b - learning_rate * db

print("Weight:", w)
print("Bias:", b)
```

```python
# Prediction on unseen input
new_x = 6
prediction = w * new_x + b
print("Predicted value:", prediction)   # should be close to 12
```

## With Vectorization (NumPy)

The inner `for i in range(m)` loop is replaced by array operations. NumPy computes all predictions and errors at once, which is shorter and much faster on large datasets.

```python
import numpy as np

# Training data as NumPy arrays
X = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([2, 4, 6, 8, 10], dtype=float)

w = 0.0
b = 0.0
learning_rate = 0.01
epochs = 1000
m = len(X)

for epoch in range(epochs):
    y_pred = w * X + b            # predictions for ALL examples at once
    error = y_pred - y            # error vector, shape (m,)

    dw = np.dot(error, X) / m     # (1/m) * sum(error * X)
    db = np.sum(error) / m        # (1/m) * sum(error)

    w = w - learning_rate * dw
    b = b - learning_rate * db

print("Weight:", w)
print("Bias:", b)
```

```python
# Prediction
new_x = 6
prediction = w * new_x + b
print("Predicted value:", prediction)
```
